```python
import tensorflow as tf
from tensorflow.keras.utils import image_dataset_from_directory
```

# Data preprocessiong

**preprocession the traning set**

```python
from google.colab import drive
drive.mount('/content/drive')
```

```python
from tensorflow.keras.preprocessing.image import ImageDataGenerator

train_datagen = ImageDataGenerator(
    rescale=1./255,
    shear_range=0.2,
    zoom_range=0.2,
    horizontal_flip=True
)

training_set = train_datagen.flow_from_directory(
    r'/content/drive/MyDrive/Colab Notebooks/dataset/training_set',
    target_size=(64, 64),
    batch_size=32,
    class_mode='binary'
)
```

**preprocessiong the test set**

```python
test_datagen=ImageDataGenerator(rescale=1./255)
test_set = test_datagen.flow_from_directory(
    r'/content/drive/MyDrive/Colab Notebooks/dataset/test_set',
    target_size=(64, 64),
    batch_size=32,
    class_mode='binary'
)
```

# **Bulding the CNN**

**Initialising the CNN**

```python
cnn=tf.keras.models.Sequential()
```

**Convolution**

```python
cnn.add(tf.keras.layers.Conv2D(filters=32,kernel_size=3,activation='relu',input_shape=[64,64,3]))
```

Pooling **bold text**

```python
cnn.add(tf.keras.layers.MaxPool2D(pool_size=2,strides=2))
```

**Adding the second convolutional layer**

```python
cnn.add(tf.keras.layers.Conv2D(filters=32,kernel_size=3,activation='relu'))
cnn.add(tf.keras.layers.MaxPool2D(pool_size=2,strides=2))
```

# **Flattening**

```python
cnn.add(tf.keras.layers.Flatten())
```

Full **conation**

```python
cnn.add(tf.keras.layers.Dense(units=128,activation='relu'))
```

**Output layers**

```python
cnn.add(tf.keras.layers.Dense(units=1,activation='sigmoid'))
```

# **Training the CNN**

**Compiling the CNN**

```python
cnn.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
```

**Traning the CNN on traning set**

```python
cnn.fit(x=training_set,validation_data=test_set,epochs=20)
```

**Making a single predication**

```python
import numpy as np
from keras.preprocessing import image
test_image=image.load_img (r'/content/drive/MyDrive/Colab Notebooks/dataset/single_predication/cat_or_dog_1',target_size=(64,64))
test_image=image.img_to_array(test_image)
test_image=np.extend_dims(test_image,axis=0)
result=cnn.predict(test_image)
training_set.class_indices
if result[0][0]==1:
 predication='Dog'
else:
  predication='cat'
print(predication)
```
